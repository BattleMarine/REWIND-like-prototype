# ArchVis v1 학습 실행
검토된 고정 소스, 실행 설정, 선택 계획, 검토 근거와 호환 raw cache가 필요합니다. 기존 v0 모델 체크포인트는 입력 캐시로 사용하지 않습니다. 실제 학습은 지원되는 CUDA GPU의 이름과 VRAM을 확인합니다. 마지막 셀에서 실행 결과를 별도로 회수합니다.

```python
from pathlib import Path
import subprocess, sys

SOURCE = Path('/content/rewind-v1-source')
RUN = Path('/content/rewind-v1-run')
RECOVERY = Path('/content/rewind-v1-recovery.zip')
RECOVERY_MAX_BYTES = None  # 고정 인계의 결과 상한
HARD_MAX_SECONDS = None  # 고정 인계의 subprocess 상한

# actual CLI의 각 필수 경로와 SHA를 고정 인계에서 그대로 전달합니다.
RUN_ARGS = []  # --config, --config-sha256, --selection, --selection-sha256, --evidence,
               # --evidence-sha256, --review-dir, --spec, --cache, --cache-sha256, --source-sha256
               # 선택 사항: --cache-local, --resume-sha256, --device
if RUN_ARGS and type(HARD_MAX_SECONDS) is int and 0 < HARD_MAX_SECONDS <= 86460:
    subprocess.run([sys.executable, '-m', 'rewind_v1', 'actual', '--output', str(RUN), *RUN_ARGS],
                   cwd=SOURCE, timeout=HARD_MAX_SECONDS, check=True)
else:
    print('검토된 실행 인계를 입력하면 시작할 수 있습니다.')
```

```python
if RUN.is_dir() and type(RECOVERY_MAX_BYTES) is int and RECOVERY_MAX_BYTES > 0:
    subprocess.run([sys.executable, '-m', 'rewind_v1', 'recover', '--run', str(RUN), '--output', str(RECOVERY),
                    '--max-bytes', str(RECOVERY_MAX_BYTES)], cwd=SOURCE, check=True)
    from google.colab import files
    files.download(str(RECOVERY))
```
